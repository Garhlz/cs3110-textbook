# 并发

并发程序让多个计算在时间上发生重叠，而不是被迫完全按顺序执行。

* 例如，*图形用户界面*（GUI）依靠并发来保持响应，同时让计算在后台继续。没有并发，GUI 就会“卡死”，直到当前操作结束。并发程序出了问题时，这种情况仍可能发生，令用户十分恼火。

* 电子表格需要并发性来重新计算所有单元格，同时保持菜单和编辑功能可用。

* Web 浏览器需要并发性来随着新数据通过网络传入而增量读取和渲染网页，
  运行嵌入在网页中的脚本，并让用户能够浏览页面、点击超链接。

*服务器*是另一类需要并发的应用程序。Web 服务器需要响应许多客户端请求，
而客户端通常不愿意等待。例如，如果作业发布在 CMS 上，你肯定希望能和
班上其他人同时查看作业，而不是像在车管所或老式熟食店那样“取号”等叫号。

*操作系统*（OS）的主要职责之一就是提供并发，让音乐播放器、浏览器、代码编辑器等多个应用可以同时运行。它有两种基本而互补的实现方式：

* **交错：** 在不同计算之间快速切换。例如，先让音乐播放器运行 100 毫秒，再运行浏览器和编辑器，然后重复这一过程。多个计算看似同时进行，实际上任一时刻只有一个计算在执行。

* **并行性：** 使用能够真正同时执行两个或多个计算的硬件。如今许多处理器都是*多核*处理器，即拥有多个中央处理单元（CPU），每个核心都能同时执行程序。

无论采用哪种方式，并发编程都极具挑战。即使多个核心可以同时工作，内存、屏幕和网络等许多资源仍须共享。既要管理共享，又不引入错误，相当困难。例如，两个程序共同使用内存时，必须协调各自何时能够读写；否则它们可能同时写入同一位置，破坏其中的数据。程序竞相赶在彼此之前完成操作所造成的这类*竞争条件*，出了名地难以避免。

最基本的挑战是，并发性会让程序执行变得*非确定性*：
操作发生的顺序不一定能提前知道。竞争条件就是非确定性的一个例子。
为了在非确定性面前正确编程，程序员不得不考虑“所有”可能的操作顺序，
并确保程序在*所有*这些顺序下都能正常工作。

纯函数式程序中的非确定性更容易推理，因为同一个表达式无论何时求值都会得到相同结果。例如，计算 `(2 * 4) + (3 * 5)` 时，左右两项乘法可以同时进行，而答案不会改变。命令式程序则麻烦得多：若同时执行 `!x` 和 `incr x; !x`，结果会随执行先后而变化。

## 线程

为了使并发编程变得更容易，计算机科学家发明了许多
抽象。最著名的之一是“线程”。抽象地讲，线程是一个
单个顺序计算。一次可以有多个线程运行，
取决于硬件，多个线程可以交错运行或并行运行；*调度器*
负责选择在任意给定时间运行哪个线程。调度可以是“抢占式”的：
调度器可以停止线程并稍后重新启动它，线程自己对此没有选择权；
也可以是*协作式*的：线程必须主动选择把控制权交还给调度器。
前者可能导致竞争条件，后者可能导致应用程序无响应。

具体来说，线程是一组加载到处理器寄存器中的值。这些值告诉处理器下一条指令在哪里、栈和堆位于内存何处，等等。为了实现抢占，调度器会在硬件中设置定时器；计时结束时，当前线程被中断，调度器随即开始运行。系统编程和操作系统课程通常会详细讲解这些概念。
